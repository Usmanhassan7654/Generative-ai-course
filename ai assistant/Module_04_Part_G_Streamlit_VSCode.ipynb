# Module 04 Part G — Convert Assistant into a Streamlit App

In Parts A–F we built the assistant in a notebook. Now we move the same code into real project files.

```text
part_g_streamlit_app/
├── app.py
├── assistant_core.py
├── business_logic.py
├── company_data.py
├── tools.py
├── requirements.txt
├── .env.example
├── .gitignore
└── data/leads.csv
```


## Step 1 — Create and activate virtual environment

```bash
cd part_g_streamlit_app
python -m venv .venv
.venv\Scripts\activate
```

Mac/Linux: `source .venv/bin/activate`

## Step 2 — Install requirements
```bash
python -m pip install --upgrade pip
pip install -r requirements.txt
```

## Step 3 — Add Groq API key
Copy `.env.example` to `.env`, then add:
```env
GROQ_API_KEY=your_key_here
```
The project uses `ChatGroq(model="openai/gpt-oss-20b")`.

## Step 4 — Understand the file split

- `company_data.py` = company information
- `business_logic.py` = normal Python functions
- `tools.py` = LangChain tools
- `assistant_core.py` = Groq LLM + system prompt + tools
- `app.py` = Streamlit chat interface

## Step 5 — Run locally
```bash
streamlit run app.py
```
Open the local URL shown in terminal, usually `http://localhost:8501`.
